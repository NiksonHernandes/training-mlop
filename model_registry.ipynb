## Recebendo o modelo registrado e suas versões

```python
registered_model = client.get_registered_model('fetal_health_nik')
```

```python
registered_model.latest_versions
```

## Obtendo o id da execução do modelo

```python
run_id = registered_model.latest_versions[-1].run_id
```

```python
run_id
```

## Carregando o modelo

```python
# logged_model = f'runs:/{run_id}/model'
# loaded_model = mlflow.pyfunc.load_model(logged_model)

# Baixa apenas o arquivo .keras
local_path = client.download_artifacts(run_id, path="model/modelo.keras", dst_path="./")
print("Salvo em:", local_path)

# Carrega com Keras diretamente
loaded_model = tf.keras.models.load_model(local_path)
print(loaded_model.summary())
```

## Fazendo uma predição com o modelo carregadO

```python
accelerations = 0
fetal_movement = 0
uterine_contractions = 0
severe_decelerations = 0
```

```python
received_data = np.array([
        accelerations,
        fetal_movement,
        uterine_contractions,
        severe_decelerations,
    ]).reshape(1, -1)
```

```python
loaded_model.predict(received_data)
```
